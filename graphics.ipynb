```python
import pandas as pd
import json

# Load the JSON data from the file
with open('Data/short_cleaned_songs.json', 'r') as f:
    data = json.load(f)

# Create a Pandas DataFrame from the JSON data
df = pd.DataFrame(data['datos_limpios'])

# Now you can work with the DataFrame 'df'
df.head()
```

```python
import seaborn as sns
import matplotlib.pyplot as plt

# Assuming 'df' is already defined as in your previous code

# Select only numeric columns for the correlation matrix
numeric_cols = df.select_dtypes(include=['number'])

# Calculate the correlation matrix
correlation_matrix = numeric_cols.corr()

# Create the heatmap using seaborn
plt.figure(figsize=(12, 10))  # Adjust figure size as needed
sns.heatmap(correlation_matrix, annot=True, cmap='coolwarm', fmt=".2f")
plt.title('Correlation Matrix of Numeric Features')
plt.show()
```

```python

genre_median_popularity = df.groupby('playlist_genre')['track_popularity'].median()

# Create a histogram showing the relationship between playlist genre and median popularity
plt.figure(figsize=(12, 6))
plt.bar(genre_median_popularity.index, genre_median_popularity.values)
plt.xlabel('Playlist Genre')
plt.ylabel('Median Track Popularity')
plt.title('Median Track Popularity per Playlist Genre')
plt.xticks(rotation=45, ha='right') # Rotate x-axis labels for better readability
plt.tight_layout() # Adjust layout to prevent labels from overlapping
plt.show()
```

```python
# Convert 'track_album_release_date' to datetime objects
df['track_album_release_date'] = pd.to_datetime(df['track_album_release_date'], errors='coerce')

# Extract the year from the release date
df['release_year'] = df['track_album_release_date'].dt.year

# Filter for the 'pop' genre
pop_df = df[df['playlist_genre'] == 'pop']

# Group by year and calculate the mean popularity
pop_evolution = pop_df.groupby('release_year')['track_popularity'].mean()

# Plot the evolution of pop music popularity over time
plt.figure(figsize=(12, 6))
plt.plot(pop_evolution.index, pop_evolution.values)
plt.xlabel('Year')
plt.ylabel('Average Track Popularity')
plt.title('Evolution of Pop Music Popularity Over Time')
plt.grid(True)
plt.show()
```
